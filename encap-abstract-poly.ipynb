```python
# to not break encapsulation and maintain private we just need to use within the class not even the sub calss
class Student:
    def __init__(self,name,rollno, age):
        self.name = name                #public
        self._rollno = rollno           #protected
        self.__age = age                # private attribute
    def __display(self):                #private method
        print(f"hi myself {self.name} from student class {self._rollno} and age is {self.__age}")      
    def displayPublic(self):
        self.__display()

s = Student(input("enter your name: "), int(input("enter your rollno: ")), int(input("Enter you age:")))
s.displayPublic() # maintaing encapsulation and accessing the private data within the class maintaing data private 
```

```python
#getters and setters
class Student:
    def __init__(self,name,age):
        self._name = name                
        self.__age = age         
    def get_age(self):          #getter to get the private attributes
        return self.__age       
    def set_age(self):          #setter to change the private attributes
        self.__age = int(input("Enter correct age: "))

b1 = Student(input("enter your name: "), int(input("enter your age: ")))
print(b1.get_age())
b1.set_age()
print(b1.get_age())
```

```python
# gettter and setter using the @property decarator
# the name of @property, @name.setter and @name.deleter methods need to be same if not rises error
class Student:
    def __init__(self):
        self.__name = "Unknown"   # Private variable
        self.__age = 0
    @property
    def name(self):        # Getter -- method name = name
        return self.__name

    @name.setter
    def name(self, value): # Setter -- method name = name
        if value:
            self.__name = value

    @name.deleter
    def name(self):        # Deleter -- method name = name
        print("Deleting name...")
        del self.__name
s = Student()
s.n = "Praveen"     # ✅ setter is called
print(s.n)          # ✅ getter is called
del s.n            # ✅ deleter is called
```

```python
#using property() function
class Person:
    def __init__(self):
        self.__name = "Unknown"

    # Getter
    def get_name(self):
        print("Getting name...")
        return self.__name

    # Setter
    def set_name(self, value):
        print("Setting name...")
        if value:
            self.__name = value
        else:
            print("❌ Name can't be empty.")

    # Deleter
    def del_name(self):
        print("Deleting name...")
        del self.__name

    # Creating the property
    name = property(get_name, set_name, del_name, "This is the name property")
p = Person()

p.name = "Praveen"   # Calls set_name()
print(p.name)        # Calls get_name()
del p.name           # Calls del_name()
print(Person.name.__doc__)  # Shows the docstring
```

```python
class Duck:
    def swim(self):
        print("i am duck and i can swim")
    def speaks(self):
        print("Quack quack")
class Dog:
    def swim(self):
        print("i am dog and i can swim")
    def speaks(self):
        print("woof woof")

def display(obj):
    obj.swim()
    obj.speaks()
    print("information displayed")

d = Duck()
display(d)      #it doesn't checks the type of object it checks whetther the method is defined or not on that object  . like the previous example which can run on datatype of int and float.

dog = Dog()
display(d)
```

```python
class DuckTyping:
    def display(self):
     self.swim()
     self.speaks()
     print("information displayed")
class Duck(DuckTyping):
    def swim(self):
        print("i am duck and i can swim")
    def speaks(self):
        print("Quack quack")
class Dog(DuckTyping):
    def swim(self):
        print("i am dog and i can swim")
    def speaks(self):
        print("woof woof")

d =Duck()
d.display()
c = Dog()
c.display()

#e =DuckTyping()
#e.display()
```

```python
import builtins # to see all the dunder methods

dunder_methods = []

for name in dir(builtins):
    obj = getattr(builtins, name)
    if isinstance(obj, type):  # Only classes
        for attr in dir(obj):
            if attr.startswith('__') and attr.endswith('__'):
                dunder_methods.append(attr)

# Remove duplicates and sort
dunder_methods = sorted(set(dunder_methods))

# Display
for method in dunder_methods:
    print(method)

print(f"\nTotal Dunder Methods: {len(dunder_methods)}")
```

```python
#method overloading and method overriding
#method overloading means using same to multiple methods with the same name but with different parameters in same class
#method overriding means using same named method and no.of parameters in parent class and child class.
```

```python
class Demo:
    def add(self,a,b):
        print(a+b)
d=Demo()
d.add(5,10)
```

```python
class Demo:
    def add(self,a,b):
        print(a+b)
    def add(self,a,b,c):
        print(a+b+c)
d=Demo()
d.add(5,10)# only recognises the last version of the method
```

```python
# to overcome use default parameters.
class Demo:
    def add(self,a,b,c=0):
        print(a+b+c)
d=Demo()
d.add(5,10)
d.add(5,10,25)
d.add(5,12,)
```

```python
# using args to overcome
class Demo:
    def add(self,*args):
        total = 0
        for i in args:
            total += i
        print(total)

d = Demo()
d.add(1,2,3,4,5)
d.add(15,56,2156,5165,54,86,56,63,778,6)
```
