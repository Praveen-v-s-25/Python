```python
# hirerachical inheritance -- combination of two or more number of inheritances
class A:
    def adisplay(self):
        print("display from class A")
class B(A):                             #single inheritance
    def bdisplay(self):
        print("display from the class B")
class C:
    def cdisplay(self):
        print("display from C class")
class D(B,C):                           #multiple inheritance
    def ddisplay(self):
        print("display from D class")

d = D()
d.adisplay()
d.bdisplay()
d.cdisplay()
d.ddisplay()

print(D.mro())
print(C.mro())
print(B.mro())
print(A.mro())
```
